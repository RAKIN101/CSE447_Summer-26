```python
x= 0b1110001100101001011
a=bin(x)
y= 0b0011000000010000001101
b=bin(y)
z= 0b0011101101111001001110
c=bin(z)

def A51(x, y, z, a):
    m = bin(x)[2:].zfill(19)
    n=bin(y)[2:].zfill(22)
    o = bin(z)[2:].zfill(23)

    keystream=""

    for i in range(a):
        count= [m[8],n[10], o[10]]
        if count.count("1") > 1:
          major = "1"
        else :
          major="0"

        if m[8]==major:
            M=str(int(m[13])^int(m[16])^int(m[17])^int(m[18]))
            m=M+m[:-1]

        if n[10]==major:
            N=str(int(n[20])^int(n[21]))
            n=N+n[:-1]

        if o[10]==major:
            O=str(int(o[7])^int(o[20])^int(o[21])^int(o[22]))
            o= O+o[:-1]

        k= str(int(m[-1])^int(n[-1])^int(o[-1]))
        keystream+=k

    return keystream


plaintext=["It is alive",
"Snap out of it",
"I am as mad as hell and I am not going to take this anymore",
"Bond James Bond",
"Love means never having to say you're sorry"]
for i in plaintext:
  print(f"Plaintext: {i}")
  n_bits=len(i)*8
  ks = A51(x, y, z, n_bits)
  print(f"keystream: {ks}")
  plaintext_bin = ''.join(format(ord(c), '08b') for c in i)

  ciphertext_bin = ''.join(str(int(p) ^ int(k)) for p, k in zip(plaintext_bin, ks))
  print(f"Ciphertext (binary): {ciphertext_bin}")

  decrypted_bin = ''.join(str(int(c) ^ int(k)) for c, k in zip(ciphertext_bin, ks))
  decrypted_text = ''.join(chr(int(decrypted_bin[i:i+8], 2)) for i in range(0, len(decrypted_bin), 8))
  print(f"Decrypted text: {decrypted_text}")
```
