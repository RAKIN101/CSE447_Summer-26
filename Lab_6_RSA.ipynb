```python
#generate large prime(128 bits) p and q
import random
import sympy
p = sympy.randprime(2**127, 2**128 - 1)
q = sympy.randprime(2**127, 2**128 - 1)
print(f"p = {p}")
print(f"q = {q}")
```

```python
#Calculate n = pq

n = p * q
print(f"n = {n}")
```

```python
#Calculate phi(n)=(p-1)(q-1)

phi_n = (p - 1) * (q - 1)
print(f"phi(n) = {phi_n}")
```

```python
#Perform encryption here and generate ciphertext [c = m^e (mod n)] and private key [d*e ≡ 1 (mod(phi(n)))]
e = 11

d = sympy.mod_inverse(e, phi_n) #e^-1 mod phi_n
c = pow(m, e, n) # m^e mod n
print(f"public_key (e, n) = ({e}, {n})")
print(f"private_key (d) = {d}")
print(f"ciphertext (c) = {c}")
```

```python
#perform decryption here [m = c^d (mod n)]

m = pow(c, d, n) # c^d mod n
print(f"decrypted_message (m) = {m}")
```
